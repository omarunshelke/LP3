```python

# Multithreaded Matrix Multiplication
import threading

A = [[1, 2],
     [3, 4]]

B = [[5, 6],
     [7, 8]]

rows_A = len(A)
cols_A = len(A[0])
cols_B = len(B[0])

C = [[0 for j in range(cols_B)]
     for i in range(rows_A)]

# Function to multiply one row
def multiply_row(i):
    for j in range(cols_B):
        for k in range(cols_A):
            C[i][j] += A[i][k] * B[k][j]

# Create threads
threads = []

for i in range(rows_A):
    t = threading.Thread(target=multiply_row, args=(i,))
    threads.append(t)
    t.start()

# Wait for all threads
for t in threads:
    t.join()

print("Multithreaded Matrix Multiplication:")

for row in C:
    print(row)
```
